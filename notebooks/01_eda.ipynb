## Summary

### 1. Data quality
- **Shape:** 7,043 customers × 21 columns. No missing values except `TotalCharges` (see below).
- **Duplicates:** no duplicate `customerID`s. Excluding the ID, 42 rows share an identical profile with at least one other row. All of them have `tenure = 1` (one bill, few possible feature combinations), so they are treated as **distinct customers with the same profile** and kept.
- **`TotalCharges`:** numeric values stored as text. 11 values are blank strings (whitespace, not `NaN`, so `isna()` does not detect them). All 11 belong to customers with `tenure = 0`, i.e. new customers who have not been billed yet, so they are **filled with 0** (a true value, not an imputation). Converted with `pd.to_numeric(errors="coerce")`, and the number of `NaN`s was verified to be exactly 11 before filling.
- **Consistency checks:** `"No internet service"` appears in all 6 internet add-on columns **if and only if** `InternetService = "No"` (1,526 customers). `"No phone service"` in `MultipleLines` matches `PhoneService = "No"` (682). The data is consistent, but these values are **redundant**: the same information is repeated in 7 columns.
- **Ranges:** no negative or implausible values. `tenure` ranges from 0 to 72 months (72 is likely a system cap). `TotalCharges` is right-skewed (mean 2,280 vs median 1,395), so medians are used for group comparisons.

### 2. Who churns: recommendations for the Retention team
Overall churn rate: **26.5%** (1,869 of 7,043).

**Highest-risk segments:**
1. **Customers in their first year:** 47.4% churn (**1.8×** the average), and the largest group (2,186 customers). After 4+ years churn drops to 9.5%.
2. **Month-to-month contracts:** 42.7% churn (**1.6×**), vs 11.3% for one-year and 2.8% for two-year contracts.
3. **Electronic check payment:** 45.3% churn (**1.7×**), vs 15–19% for the other payment methods.

**Other notable signals:** Fiber optic internet (41.9%), no tech support (41.6%), no online security (41.8%), senior citizens (41.7%), no dependents (31.3%), paperless billing (33.6%).

**Little or no relationship with churn:** gender (26.9% vs 26.2%), phone service, multiple lines, and streaming services (~30% vs ~34% among internet users).

**Numeric features (median, churned vs retained):**
- `tenure`: **10 vs 38 months**, the strongest difference in the dataset.
- `MonthlyCharges`: **79.65 vs 64.43** (churned customers pay ~24% more per month).
- `TotalCharges`: 704 vs 1,680. This is a consequence of shorter tenure, not an independent finding.

**Caveats:**
- These segments **overlap heavily**: a first-year, month-to-month, electronic-check, fiber customer is often the same person. Segment sizes cannot be added up.
- These are **correlations, not causes**.

### 3. Features for modelling (input to MLP-1)
- **Target:** `Churn` (Yes/No → 1/0). Classes are imbalanced (26.5% positive), so use a **stratified** train/test split and evaluate with ROC AUC rather than accuracy.
- **Categorical (one-hot):** `gender`, `Partner`, `Dependents`, `PhoneService`, `MultipleLines`, `InternetService`, `OnlineSecurity`, `OnlineBackup`, `DeviceProtection`, `TechSupport`, `StreamingTV`, `StreamingMovies`, `Contract`, `PaperlessBilling`, `PaymentMethod`.
- **Binary, already encoded:** `SeniorCitizen` (0/1, pass through unchanged).
- **Numeric:** `tenure`, `MonthlyCharges`, `TotalCharges`.
- **Dropped:** `customerID` (identifier, no predictive meaning).
- **Preprocessing must live in the pipeline, not the notebook:** the API will receive raw values (e.g. `"Yes"`, `" "` in `TotalCharges`), so type conversion, filling and encoding have to be part of the saved model artifact.

**Risks to address in MLP-1:**
- **`TotalCharges` multicollinearity:** r = 0.83 with `tenure` and 0.65 with `MonthlyCharges` (≈ tenure × monthly charge). Decide whether to keep it for a logistic regression.
- **Redundant service categories:** `"No internet service"` / `"No phone service"` duplicate `InternetService` / `PhoneService`. Consider mapping them to `"No"`.
- `tenure_group` was created for EDA only and is **not** a model feature.
